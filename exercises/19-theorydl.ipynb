## Exercise 19.8: Reproducing Double Descent with Polynomial Regression

The chapter's own double descent demonstration uses Python; here we reproduce the same phenomenon in R. Simulate $n=20$ noisy observations $y_i = f(x_i) + \varepsilon_i$ of $f(x)=\sin(5x)$ at equally spaced points $x_i \in [-1,1]$, with $\varepsilon_i \sim N(0, 0.3^2)$, and fit polynomial regressions of degree $d=1,\ldots,45$ using the standard Legendre polynomials $P_0,\ldots,P_d$ (with $P_k(1)=1$) as the basis. Measure the test error as the RMSE between the fitted polynomial and $f$ on a grid of 200 equally spaced points in $[-1,1]$. Because $d$ ranges well past $n$, use the Moore-Penrose pseudoinverse (`MASS::ginv`) rather than ordinary least squares, so that the overparameterized regime returns the minimum-norm interpolating solution instead of an error. This is exactly the classical-U-then-re-descent pattern documented for over-parameterized models (Belkin et al. 2019; Hastie et al. 2022).

a) Carry out the simulation with `set.seed(42)`. Report the degree with the smallest test RMSE in the classical (underparameterized) regime, the degree at which test RMSE peaks, and the test RMSE at degree 45.

b) Explain why the training RMSE becomes (numerically) zero once $d$ reaches a certain value, while the test RMSE does not.
