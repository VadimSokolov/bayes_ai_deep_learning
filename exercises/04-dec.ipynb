## Exercise 4.11: Certainty Equivalent and Risk Premium

Return to the fundraiser dinner of the "Decisions" exercise above. Attendance $X$ takes values $100, 200, 300, 400, 500$ with probabilities $0.1, 0.2, 0.3, 0.2, 0.2$, and profit under the flat-fee rental scheme is $Y_1 = 40X - 1500$, with $\mathrm{E}\left(Y_1\right) = 11{,}300$. A risk-neutral organizer values the dinner at $\mathrm{E}\left(Y_1\right)$, but a risk-averse organizer values it less. Suppose the organizer's preferences follow a CARA (constant absolute risk aversion) utility function
$$
u(w) = \frac{1 - e^{-aw}}{a}, \qquad a > 0,
$$
where the coefficient $a = -u''(w)/u'(w)$ is the (constant) coefficient of absolute risk aversion.

a) Write an R function for $u(w,a)$ and compute the expected utility $\mathrm{E}\left(u(Y_1)\right)$ for $a = 0.0002$.

$$
CE = -\frac{1}{a}\log\left(1 - a\,\mathrm{E}\left(u(Y_1)\right)\right),
$$
and compute $CE$ and the risk premium $\mathrm{E}\left(Y_1\right) - CE$ for $a = 0.0002$.

c) Compute $CE(a)$ over a grid of $a$ from near $0$ to $0.0015$ and plot the risk premium $\mathrm{E}\left(Y_1\right) - CE(a)$ against $a$. What happens to the risk premium as $a \to 0$? As $a$ grows, what value does $CE$ approach, and why does that make sense?

## Exercise 4.12: The Kelly Criterion and Long-Run Growth

a) Show that the expected log-growth rate of wealth per round is
$$
G(f) = p\log(1+bf) + q\log(1-f),
$$
and, by solving $G'(f)=0$, derive the optimal (full-Kelly) fraction $f^\star = p - q/b$. Verify $G''(f)<0$ everywhere, confirming this is a maximum, and evaluate $f^\star$ numerically for $p=0.4, b=2$.

b) In R, define `G(f)` and confirm numerically, by a grid search or `optimize()`, that its maximizer matches your formula for $f^\star$.

c) Simulate $N=5{,}000$ independent paths of $T=200$ bets each, starting from wealth $W_0=1$, for three betting fractions: full Kelly $f^\star$, half Kelly $f^\star/2$, and an over-betting fraction $3f^\star$. Report, for each fraction, the median terminal wealth, the median terminal log-wealth, and the *ruin frequency*, defined as the fraction of paths whose wealth ever drops below $1\%$ of $W_0$. Set a seed.

d) Compare $\text{median}(\log W_T)/T$ from your simulation to the theoretical $G(f)$ from part (a) for each fraction. Discuss the trade-off between growth rate and ruin risk as you move from half-Kelly to full-Kelly to over-betting.
