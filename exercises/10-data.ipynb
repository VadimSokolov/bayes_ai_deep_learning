## Exercise 10.7: Simulating Galton's Ox

Galton's crowd of 800 people estimated an ox weighing 1,198 lbs with a median guess of 1,207 lbs.

a) Simulate 800 guesses from a $N(1200, 100^2)$ distribution (slight bias, moderate noise). Compute the mean, median, and trimmed mean (10% trimmed). Which is closest to 1,198?

b) Now add 20 outliers (replace 20 guesses with values drawn from $N(2000, 200^2)$). Recompute mean, median, and trimmed mean. Which aggregation method is most robust?

c) Repeat the experiment (without outliers) 1000 times. Plot the distribution of medians. What is its standard deviation compared to the standard deviation of a single guess?
