## Exercise 6.8: Sequential Bayesian Testing

Alan Turing used sequential Bayesian testing to crack the Enigma code. Suppose you're testing whether a substitution cipher maps letter A to letter X. Enigma traffic was in German, so this exercise uses German letter statistics (the chapter's Enigma example, Example 6.2, uses English for illustration).

- Under $H_0$ (correct mapping): $P(\text{match}) = 0.08$, an aggregate match/coincidence probability for German text (analogous to the index-of-coincidence statistic $m \approx 0.066$ the chapter computes for English), not a single-letter frequency
- Under $H_1$ (incorrect mapping): $P(\text{match}) = 1/26 \approx 0.0385$

You observe a sequence of positions where you expect letter A, and record whether you see X.

a) What is the Bayes factor for a single match observation?

b) What is the Bayes factor for a single non-match observation?

c) If you start with prior odds of 1:25 (one correct mapping among 26 possibilities), how many consecutive matches do you need to achieve posterior odds of 10:1?

d) Plot the evolution of log posterior odds as you observe: match, match, non-match, match, match, match.

## Exercise 6.10: Posterior Odds with Asymmetric Losses

A security system must decide whether to flag a transaction as fraudulent. Let $H_0$ be "legitimate" and $H_1$ be "fraudulent."

- Loss matrix: Missing fraud ($d_0 \mid H_1$) costs \$10,000; False alarm ($d_1 \mid H_0$) costs \$100
- Prior probability of fraud: $P(H_1) = 0.01$
- Based on transaction features, the Bayes factor is $BF_{10} = 5$

a) Calculate the posterior probability of fraud.

b) What is the optimal decision rule based on minimizing expected loss?

c) Should this transaction be flagged?

d) What Bayes factor threshold would lead to flagging the transaction?
