## Exercise 18.8: ReLU vs Tanh on a Smooth Target

The chapter fits a one-hidden-layer network (3 hidden units, shared hidden bias, single linear output: 8 parameters total) to a noisy quadratic function by minimizing the sum of squared errors with `optim(..., method = "BFGS")`. Here you will compare the ReLU and $\tanh$ hidden activations on a more curved target.

Simulate training data from
$$
y = \sin(2x) + \epsilon, \quad \epsilon \sim N(0, 0.05^2), \quad x \in [-1, 1]
$$

a) Using the chapter's network (one hidden layer, 3 units, single linear output), fit the model with the ReLU hidden activation. Minimize the sum of squared errors with `optim(..., method = "BFGS")` and report the final training loss.

b) Fit the same architecture with the $\tanh$ hidden activation, using the same loss and optimizer. Report the final training loss.

c) Plot the training data together with both fitted curves over $x \in [-1, 1]$.

d) Which activation achieves the lower training loss? Comment on the smoothness of the two fits: the ReLU network is piecewise linear (look for visible kinks where hidden units switch on or off), while the $\tanh$ network is smooth and differentiable everywhere.
